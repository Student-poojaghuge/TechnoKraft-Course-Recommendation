```python
df_model = df.drop(
    columns=["Course", "Competitors_Offer"]
)

df_model.head()
```

```python
x=df_model.drop(columns=["Recommendation"])
y=df_model["Recommendation"]
```

```python
print("Features:")
print(x.columns)

print("\nTarget:")
print(y.name)
```

```python
X_encoded = pd.get_dummies(
    x,
    columns=[
        "Category",
        "Market_Demand",
        "Competition",
        "TechnoKraft_Offers",
        "Duration",
        "Placement_Demand",
        "AI_Trend"
    ],
    dtype=int
)

X_encoded.head()
```

```python
y_encoded = y.map({
    "No": 0,
    "Yes": 1
})
```

```python
y_encoded.value_counts()
```

```python
print("X shape:", X_encoded.shape)
print("y shape:", y_encoded.shape)
```

```python
print(X_encoded.dtypes)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

y_pred
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy (%):", accuracy * 100)
```

```python
from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    y_pred,
    target_names=["No", "Yes"]
))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)

print("Model Performance")
print("------------------")
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")
```

```python
print(f"Accuracy  : {accuracy * 100:.2f}%")
print(f"Precision : {precision * 100:.2f}%")
print(f"Recall    : {recall * 100:.2f}%")
print(f"F1 Score  : {f1 * 100:.2f}%")
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=["Not Recommended", "Recommended"],
    yticklabels=["Not Recommended", "Recommended"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")

plt.show()
```

```python
technokraft_name = "TechnoKraft Training & Solution Pvt. Ltd."

technokraft_df = institute_df[
    institute_df["Institute"] == technokraft_name
]

competitor_df = institute_df[
    institute_df["Institute"] != technokraft_name
]

print("TechnoKraft records:", len(technokraft_df))
print("Competitor records:", len(competitor_df))
```

```python
common_courses = sorted(
    set(technokraft_df["Course"]) &
    set(competitor_df["Course"])
)

print("Common courses:", len(common_courses))
print(common_courses)
```

```python
comparison_df = competitor_df[
    competitor_df["Course"].isin(common_courses)
].groupby("Course").agg(
    Competitor_Min_Fee=("Min_Fee_INR", "mean"),
    Competitor_Max_Fee=("Max_Fee_INR", "mean"),
    Competitor_Avg_Fee=("Avg_Fee_INR", "mean")
).reset_index()

comparison_df
```

```python
technokraft_common = technokraft_df[
    technokraft_df["Course"].isin(common_courses)
][
    ["Course", "Min_Fee_INR", "Max_Fee_INR", "Avg_Fee_INR"]
].sort_values("Course")

technokraft_common
```

```python
fee_comparison = comparison_df.merge(
    technokraft_common,
    on="Course"
)

fee_comparison = fee_comparison.rename(columns={
    "Min_Fee_INR": "TechnoKraft_Min_Fee",
    "Max_Fee_INR": "TechnoKraft_Max_Fee",
    "Avg_Fee_INR": "TechnoKraft_Avg_Fee"
})

fee_comparison
```

```python
fee_comparison["Suggested_TechnoKraft_Fee"] = (
    fee_comparison["Competitor_Avg_Fee"].round(0)
)

fee_comparison[
    [
        "Course",
        "Competitor_Avg_Fee",
        "TechnoKraft_Avg_Fee",
        "Suggested_TechnoKraft_Fee"
    ]
]
```

```python
fee_comparison["Fee_Difference"] = (
    fee_comparison["Suggested_TechnoKraft_Fee"]
    - fee_comparison["TechnoKraft_Avg_Fee"]
)

fee_comparison[
    [
        "Course",
        "TechnoKraft_Avg_Fee",
        "Suggested_TechnoKraft_Fee",
        "Fee_Difference"
    ]
]
```

```python
final_fee_comparison = fee_comparison[
    [
        "Course",
        "Competitor_Min_Fee",
        "Competitor_Max_Fee",
        "Competitor_Avg_Fee",
        "TechnoKraft_Avg_Fee",
        "Suggested_TechnoKraft_Fee",
        "Fee_Difference"
    ]
].copy()

final_fee_comparison
```

```python
final_fee_comparison.to_csv(
    "../data/TechnoKraft_Fee_Comparison.csv",
    index=False
)

print("Fee comparison saved successfully.")
```
